```python
from mpmath import mp
import pandas as pd
import numpy as np

# Set the required number of digits (e.g., 500)
mp.dps = 500
pi_digits = str(mp.pi).replace('.', '')
n = len(pi_digits)
```

```python
# Constants
phi = (1 + 5**0.5) / 2
golden_angle = 2 * np.pi / (phi**2)
step_angle_default = 0.15
step_radius = 0.5
max_radius = n * step_radius

data = []

# Generate data for both modes
for mode in ['Original', 'Golden Angle']:
    for i in range(n):
        # Select angle based on the current mode
        current_angle = golden_angle if mode == 'Golden Angle' else step_angle_default
        
        angle = -(i * current_angle)
        radius = max_radius - (i * step_radius)
        
        x = radius * np.cos(angle)
        y = radius * np.sin(angle)
        
        # Add 'Type' column for filtering in Tableau
        data.append([pi_digits[i], i, x, y, mode])

df = pd.DataFrame(data, columns=['Digit', 'Position', 'X', 'Y', 'Type'])
df.to_csv('./pi_spiral_combined.csv', index=False)
print("File ready for Tableau")
```

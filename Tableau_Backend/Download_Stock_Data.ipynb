```python
import yfinance as yf
import pandas as pd
```

```python
# 1. Settings: List of companies and time period
tickers = ['ASML','MU','SNDK','WDC']  # Add required tickers here
start_date = "2025-01-01"
end_date = "2026-03-01"

all_data = []

for ticker in tickers:
    print(f"Downloading {ticker}...")
    # Download data for one ticker at a time
    df_temp = yf.download(ticker, start=start_date, end=end_date)
    
    if not df_temp.empty:
        # Reset index (Date becomes a column)
        df_temp = df_temp.reset_index()
        
        # Add a column with the ticker name
        df_temp['Ticker'] = ticker
        
        # Rename columns (remove multi-index if present)
        # In new yf versions, columns might be named ('Open', 'ASML')
        if isinstance(df_temp.columns, pd.MultiIndex):
            df_temp.columns = df_temp.columns.get_level_values(0)

        # Select and rename only required columns
        df_temp = df_temp[['Date', 'Ticker', 'Open', 'Low', 'High', 'Close']]
        df_temp.columns = ['Date', 'Ticker', 'Price_Open', 'Price_Low', 'Price_High', 'Price_Close']
        
        all_data.append(df_temp)

# Combine everything into a single DataFrame
final_df = pd.concat(all_data, ignore_index=True)

# Save to CSV
final_df.to_csv("stocks_data.csv", index=False)

print("File stocks_data.csv created successfully!")
print(final_df.head())
```
